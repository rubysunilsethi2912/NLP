```python
# data cleaning
def clean_text(text):
    text = str(text).lower()
    
    # Remove URLs
    text = re.sub(r"http\S+|www\S+|https\S+", "", text)
    
    # Remove mentions
    text = re.sub(r"@\w+", "", text)
    
    # Remove hashtags symbol but keep the word
    text = re.sub(r"#", "", text)
    
    # Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))
    
    # Remove numbers
    text = re.sub(r"\d+", "", text)
    
    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()
    
    return text
```

```python
# Text cleaning
def clean_text(text):

    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r"http\S+|www\S+|https\S+", " ", text)

    # Remove mentions
    text = re.sub(r"@\w+", " ", text)

    # Remove hashtag symbol
    text = re.sub(r"#", " ", text)

    # Remove special characters and numbers
    text = re.sub(r"[^a-z\s]", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
# Top words
from collections import Counter
def get_top_words(text_series, n=20):

    words = []

    for text in text_series:
        words.extend(text.split())

    counter = Counter(words)

    return counter.most_common(n)
```

```python
X = df["clean_text"]

y = df["target"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", len(X_train))
print("Testing data:", len(X_test))
```

```python
# TF-IDF Feature Engineering
tfidf = TfidfVectorizer(
    ngram_range=(1,2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True,
    max_features=50000
)
```

```python
# Logistic Regression 
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
logistic_model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            ngram_range=(1,2),
            min_df=2,
            max_df=0.95,
            sublinear_tf=True,
            max_features=50000
        )
    ),
    (
        "classifier",
        LogisticRegression(
            max_iter=1000,
            C=2
        )
    )
])

logistic_model.fit(
    X_train,
    y_train
)

print("Logistic Regression training completed!")
```

```python
# Naive Bayes

from collections import Counter
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
nb_model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            ngram_range=(1,2),
            min_df=2,
            max_df=0.95,
            sublinear_tf=True,
            max_features=50000
        )
    ),
    (
        "classifier",
        MultinomialNB(alpha=0.1)
    )
])

nb_model.fit(X_train, y_train)

print("Naive Bayes training completed!")
```

```python
# Linear SVM
from sklearn.svm import LinearSVC
svm_model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            ngram_range=(1,2),
            min_df=2,
            max_df=0.95,
            sublinear_tf=True,
            max_features=50000
        )
    ),
    (
        "classifier",
        LinearSVC(C=1)
    )
])

svm_model.fit(X_train, y_train)

print("Linear SVM training completed!")
```

```python
# Model comparison
models = {
    "Logistic Regression": logistic_model,
    "Naive Bayes": nb_model,
    "Linear SVM": svm_model
}

results = []

for name, model in models.items():

    prediction = model.predict(X_test)

    accuracy = accuracy_score(
        y_test,
        prediction
    )

    precision = precision_score(
        y_test,
        prediction
    )

    recall = recall_score(
        y_test,
        prediction
    )

    f1 = f1_score(
        y_test,
        prediction
    )

    results.append([
        name,
        accuracy,
        precision,
        recall,
        f1
    ])

results_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ]
)

results_df
```

```python
# MOdel Comparison
x = np.arange(len(results_df))

width = 0.2

plt.figure(figsize=(6,4))

plt.bar(
    x - 1.5*width,
    results_df["Accuracy"],
    width,
    label="Accuracy"
)

plt.bar(
    x - 0.5*width,
    results_df["Precision"],
    width,
    label="Precision"
)

plt.bar(
    x + 0.5*width,
    results_df["Recall"],
    width,
    label="Recall"
)

plt.bar(
    x + 1.5*width,
    results_df["F1 Score"],
    width,
    label="F1 Score"
)

plt.xticks(
    x,
    results_df["Model"],
    rotation=15
)

plt.ylim(0,1)

plt.ylabel("Score")
plt.title("Model Performance Comparison")

plt.legend()

plt.show()
```

```python
# Classification Report 
best_prediction = logistic_model.predict(X_test)

print(
    classification_report(
        y_test,
        best_prediction,
        target_names=[
            "Non-Disaster",
            "Disaster"
        ]
    )
)
```

```python
# Confusion Matrix 
cm = confusion_matrix(
    y_test,
    best_prediction
)

plt.figure(figsize=(6,4))

plt.imshow(cm)

plt.title("Confusion Matrix")

plt.colorbar()

plt.xticks(
    [0,1],
    ["Non-Disaster", "Disaster"]
)

plt.yticks(
    [0,1],
    ["Non-Disaster", "Disaster"]
)

for i in range(2):

    for j in range(2):

        plt.text(
            j,
            i,
            cm[i,j],
            ha="center",
            va="center",
            fontsize=16
        )

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
# ROC Curve
probabilities = logistic_model.predict_proba(
    X_test
)[:,1]

fpr, tpr, thresholds = roc_curve(
    y_test,
    probabilities
)

roc_auc = auc(
    fpr,
    tpr
)

print("ROC-AUC:", roc_auc)
```

```python
plt.figure(figsize=(4,2))

plt.plot(
    fpr,
    tpr,
    label=f"AUC = {roc_auc:.3f}"
)

plt.plot(
    [0,1],
    [0,1],
    "--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve")

plt.legend()

plt.show()
```

```python
# Precision Recall Curve
from sklearn.metrics import precision_recall_curve, average_precision_score
precision, recall, thresholds = precision_recall_curve(
    y_test,
    probabilities
)

average_precision = average_precision_score(
    y_test,
    probabilities
)

print(
    "Average Precision:",
    average_precision
)
```

```python
plt.figure(figsize=(4,2))

plt.plot(
    recall,
    precision,
    label=f"AP = {average_precision:.3f}"
)

plt.xlabel("Recall")
plt.ylabel("Precision")

plt.title("Precision-Recall Curve")

plt.legend()

plt.show()
```

```python
# Cross Validation
from sklearn.model_selection import cross_val_score
cv_scores = cross_val_score(
    logistic_model,
    X,
    y,
    cv=5,
    scoring="f1"
)

print("Cross-validation scores:")

print(cv_scores)

print(
    "Mean F1 Score:",
    cv_scores.mean()
)

print(
    "Standard Deviation:",
    cv_scores.std()
)
```

```python
plt.figure(figsize=(4,2))

plt.bar(
    ["Fold 1", "Fold 2", "Fold 3", "Fold 4", "Fold 5"],
    cv_scores
)

plt.ylim(0,1)

plt.title("5-Fold Cross Validation")

plt.xlabel("Fold")
plt.ylabel("F1 Score")

plt.show()
```

```python
# Final Model save
final_model = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            ngram_range=(1,2),
            min_df=2,
            max_df=0.95,
            sublinear_tf=True,
            max_features=50000
        )
    ),
    (
        "classifier",
        LogisticRegression(
            max_iter=1000,
            C=2
        )
    )
])

final_model.fit(
    X,
    y
)

joblib.dump(
    final_model,
    "disaster_tweet_model.pkl"
)

print("Model saved successfully!")
```
